```python
!pip install yfinance feedparser

import yfinance as yf
import feedparser

def daten_agent(ticker):
    kurse = yf.Ticker(ticker).history(period="1mo")
    return kurse[["Close"]].tail(5)

def news_agent():
    feed = feedparser.parse("https://feeds.bbci.co.uk/news/world/rss.xml")
    return [eintrag.title for eintrag in feed.entries[:5]]

print(daten_agent("AAPL"))
print(news_agent())
```

```python
!pip install anthropic

from google.colab import userdata
import anthropic

client = anthropic.Anthropic(api_key=userdata.get("ANTHROPIC_API_KEY"))

def makro_agent(schlagzeilen):
    text = "\n".join(schlagzeilen)
    antwort = client.messages.create(
        model="claude-sonnet-5-5",
        max_tokens=800,
        messages=[{"role": "user", "content":
            "Du bist ein Makro-Analyst. Bewerte diese Weltnachrichten: "
            "Welche Branchen und Aktien könnten profitieren oder leiden? "
            "Antworte kurz auf Deutsch.\n\n" + text}]
    )
    return "".join(teil.text for teil in antwort.content if hasattr(teil, "text"))

print(makro_agent(news_agent()))
```
