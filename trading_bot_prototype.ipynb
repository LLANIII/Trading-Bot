```python
import requests
from datetime import datetime

# =====================================
# AI TRADING BOT
# BOT 1 – MARKTANALYSE
# VERSION 1.1
# =====================================

def markt_analyse():

    print("🤖 AI TRADING BOT")
    print("📊 MARKTANALYSE")
    print("-----------------------------")

    url = "https://api.coingecko.com/api/v3/simple/price"

    parameter = {
        "ids": "bitcoin,ethereum",
        "vs_currencies": "eur",
        "include_24hr_change": "true"
    }

    try:
        antwort = requests.get(
            url,
            params=parameter,
            timeout=15
        )

        antwort.raise_for_status()
        daten = antwort.json()

        for coin in ["bitcoin", "ethereum"]:

            if coin not in daten:
                print(f"⚠️ Keine Daten für {coin}")
                continue

            preis = daten[coin].get("eur")
            veraenderung = daten[coin].get("eur_24h_change")

            if preis is None:
                print(f"⚠️ Kein Preis für {coin}")
                continue

            print(f"\n🪙 {coin.upper()}")
            print(f"💰 Preis: {preis:,.2f} EUR")

            if veraenderung is not None:
                print(
                    f"📈 Veränderung 24h: "
                    f"{veraenderung:+.2f}%"
                )

        print("\n🕒 Zeitpunkt:", datetime.now())
        print("✅ Analyse abgeschlossen")

    except requests.exceptions.HTTPError as fehler:
        print("❌ API-Fehler:", fehler)
        print("HTTP-Status:", antwort.status_code)

    except requests.exceptions.RequestException as fehler:
        print("❌ Verbindungsfehler:", fehler)

    except (ValueError, TypeError) as fehler:
        print("❌ Datenfehler:", fehler)


# BOT STARTEN
markt_analyse()
```
